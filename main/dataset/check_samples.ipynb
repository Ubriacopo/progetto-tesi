```python
# Check how long samples are on average and median
# We check each dataset independently
datasets = data.train_dataset.datasets
iters = [iter(ds) for ds in datasets]
```

```python
import torch

common_eeg_channel = 2

eeg_times = []
vid_times = []

for i in iters:
    for batch in i:
        for entry in range(len(batch["student/eeg/mask"])):
            eeg_times.append(batch["student/eeg/mask"][entry].T[common_eeg_channel].sum())
            vid_times.append(batch["student/vid/mask"][entry].sum() * 4)

    eeg_times = torch.tensor(eeg_times)
    print("Dataset reports for ds:")
    print(eeg_times.float().mean())
    print((eeg_times > 16).sum())
    print(eeg_times.size())
    print("------------------")

    eeg_times = []
    vid_times = []
```
