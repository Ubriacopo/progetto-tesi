```python
from collections import Counter
import torch

counter = Counter()

for y in dm.train_dataset:
    if torch.isnan(y["assessment", "scores"]).any():
        continue
    counter[int(y["assessment", "scores"] >= 5.)] += 1

print(counter)
```
