```python
import torch
import math
scores_value = 0
score_count = 0
for i in it:
    if not math.isnan(i["assessment", "scores"].sum().item()):
        scores_value += i["assessment", "scores"].sum().item()
        score_count += i.batch_size[0]
```

```python
baseline = scores_value /score_count
baseline
```

```python
dl = dm.test_dataloader()
it = iter(dl)

all_y = []
for i in it:
    all_y.append(i["assessment", "scores"])

y = torch.cat(all_y)

baseline_mse = ((y - baseline) ** 2).mean()
baseline_rmse = baseline_mse.sqrt()
baseline_mae = (y - baseline).abs().mean()
```
