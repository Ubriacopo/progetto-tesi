```python
import numpy as np
```

# With MoCo

```python
# Overall with MoCo training is pretty unstable
baseline = np.array([0.6480, 0.6336, 0.7105])  # For Seeds: [1, 42, 150]

baseline_mean = baseline.mean().item()
baseline_std = baseline.std().item()
print("mean:", baseline_mean, " - std:", baseline_std)
```

```python
kd_mrr = np.array([0.67, 0.6505])  # For seeds [42, 150]
(kd_mrr - baseline[1:]).mean()
```

Delta is small and inconsistent and we cant say if the use of KD as harmful. <br>
It also doesn't tell us that it helps the training objective

# Without MoCo
MoCo was ablated

```python
baseline = np.array([0.7714, 0.7863])  # For Seeds: [42, 150]

baseline_mean = baseline.mean().item()
baseline_std = baseline.std().item()
print("mean:", baseline_mean, " - std:", baseline_std)
```

```python
kd_mrr = np.array([0.7666])  # For seeds: [150]
```

```python
(kd_mrr - baseline[1:]).mean().item()
```

```python
# For now, we know that KD helps the model learn.
# More experiments to assess how much would be needed.
# It is 2 times the std of baseline.
```
