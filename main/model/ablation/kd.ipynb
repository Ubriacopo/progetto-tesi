# With MoCo

```python
import numpy as np

# Overall with MoCo training is pretty unstable
baselines = np.array([0.6480, 0.6336, 0.7105])  # For Seeds: [1, 42, 150]

baseline_mean = baselines.mean().item()
baseline_std = baselines.std().item()
print("mean:", baseline_mean, " - std:", baseline_std)
```

```python
kd_mrr = np.array([0.67, 0.6505])  # For seeds [42, 150]
(kd_mrr - baselines[1:]).mean()
```

Delta is small and inconsistent and we cant say if the use of KD as harmful. <br>
It also doesn't tell us that it helps the training objective
